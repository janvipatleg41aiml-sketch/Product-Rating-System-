# E-Commerce Product Rating Prediction
### Regression ML Project

Target: **Product_Rating**

Models: Linear Regression and Random Forest Regression

Evaluation: MAE, MSE and RMSE

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
#load the dataset
from google.colab import files
import io

uploaded = files.upload()

filename = next(iter(uploaded))

df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("Dataset loaded successfully!")
print("File name:", filename)

df.head()
```

```python
#understand the dATASHAPE
print('Dataset Shape:', df.shape)

print('\nColumn Names:')
print(df.columns.tolist())

print('\nData Types:')
print(df.dtypes)
```

## 3. Statistical Summary

```python
df.describe()
```

## 4. Check Missing Values

```python
print('Missing values before cleaning:')
print(df.isnull().sum())
```

## 5. Handle Missing Values

```python
# Fill missing numerical values with the median
df = df.fillna(df.median(numeric_only=True))

print('Missing values after cleaning:')
print(df.isnull().sum())
```

## 6. EDA - Product Rating Distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df['Product_Rating'], bins=20)
plt.xlabel('Product Rating')
plt.ylabel('Number of Products')
plt.title('Distribution of Product Ratings')
plt.show()
```

## 7. EDA - Correlation Heatmap

```python
correlation = df.corr(numeric_only=True)
plt.figure(figsize=(10,7))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

## 8. Define Features (X) and Target (y)

```python
# Product_Rating is the target we want to predict
y = df['Product_Rating']

# All remaining columns are input features
X = df.drop('Product_Rating', axis=1)

print('Features:')
print(X.columns.tolist())
print('\nFeature Shape:', X.shape)
print('Target Shape:', y.shape)
```

## 9. Convert Features to Numerical Representation

```python
# The uploaded dataset already contains numerical features.
# Convert/verify all feature columns as numeric.
X = X.apply(pd.to_numeric, errors='coerce')

# Handle any values that became missing during conversion
X = X.fillna(X.median())

print(X.dtypes)
```

## 10. Split Dataset into Training and Testing Data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print('Training Features:', X_train.shape)
print('Testing Features:', X_test.shape)
print('Training Target:', y_train.shape)
print('Testing Target:', y_test.shape)
```

## 11. Regression Model 1 - Linear Regression

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
print('Linear Regression model trained successfully.')
```

## 12. Predict Ratings - Linear Regression

```python
y_pred_linear = linear_model.predict(X_test)

print('First 10 Actual Ratings:')
print(y_test.values[:10])

print('\nFirst 10 Predicted Ratings:')
print(y_pred_linear[:10])
```

## 13. Evaluate Linear Regression

```python
mae_linear = mean_absolute_error(y_test, y_pred_linear)
mse_linear = mean_squared_error(y_test, y_pred_linear)
rmse_linear = np.sqrt(mse_linear)

print('Linear Regression Results')
print('-------------------------')
print('MAE :', mae_linear)
print('MSE :', mse_linear)
print('RMSE:', rmse_linear)
```

## 14. Regression Model 2 - Random Forest Regression

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)
print('Random Forest Regression model trained successfully.')
```

## 15. Predict Ratings - Random Forest

```python
y_pred_rf = rf_model.predict(X_test)

print('First 10 Actual Ratings:')
print(y_test.values[:10])

print('\nFirst 10 Predicted Ratings:')
print(y_pred_rf[:10])
```

## 16. Evaluate Random Forest

```python
mae_rf = mean_absolute_error(y_test, y_pred_rf)
mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)

print('Random Forest Results')
print('---------------------')
print('MAE :', mae_rf)
print('MSE :', mse_rf)
print('RMSE:', rmse_rf)
```

## 17. Compare Models in a Tabular Column

```python
comparison = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest Regression'],
    'MAE': [mae_linear, mae_rf],
    'MSE': [mse_linear, mse_rf],
    'RMSE': [rmse_linear, rmse_rf]
})

comparison
```

## 18. Determine the Model with Lower Error

```python
best_model = comparison.loc[comparison['RMSE'].idxmin(), 'Model']
best_rmse = comparison['RMSE'].min()

print('Model with the lowest RMSE:', best_model)
print('RMSE:', best_rmse)
print('\nLower MAE, MSE and RMSE indicate lower prediction error.')
```

## 19. Actual vs Predicted Ratings

```python
results = pd.DataFrame({
    'Actual_Rating': y_test.values,
    'Linear_Regression': y_pred_linear,
    'Random_Forest': y_pred_rf
})

results.head(10)
```

## 20. Predict Rating for a New Product

```python
# Example values for a new product
new_product = pd.DataFrame({
    'Product_Price': [200],
    'Discount_Rate': [0.20],
    'Number_of_Reviews': [500],
    'Stock_Availability': [1],
    'Days_to_Deliver': [10],
    'Return_Rate': [0.15],
    'Category_ID': [5]
})

predicted_rating = rf_model.predict(new_product)
print('Predicted Product Rating:', predicted_rating[0])
```
